```
using System;
using System.Collections.Generic;
using System.Linq;

public class Book
{
    private string title;
    private string author;
    private int yearOfPublication;
    private string isbn;
    private string publisher;
    private int pages;
    private string language;
    
    public string Condition { get; set; } = "Хорошее";
    public int InventoryNumber { get; set; }
    public string ShelfLocation { get; set; } = "Не указано";
    public DateTime DateAdded { get; set; } = DateTime.Now;

    public delegate void BookConditionHandler(string message);
    public event BookConditionHandler ConditionChanged;

    public string Title
    {
        get { return title; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название книги не может быть пустым.");
            title = value;
        }
    }

    public string Author
    {
        get { return author; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Автор книги не может быть пустым.");
            author = value;
        }
    }

    public int YearOfPublication
    {
        get { return yearOfPublication; }
        set
        {
            if (value <= 0 || value > DateTime.Now.Year)
                throw new ArgumentException("Некорректный год публикации.");
            yearOfPublication = value;
        }
    }

    public string Isbn
    {
        get { return isbn; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("ISBN не может быть пустым.");
            isbn = value;
        }
    }

    public string Publisher
    {
        get { return publisher; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Издательство не может быть пустым.");
            publisher = value;
        }
    }

    public int Pages
    {
        get { return pages; }
        set
        {
            if (value <= 0)
                throw new ArgumentException("Количество страниц должно быть больше 0.");
            pages = value;
        }
    }

    public string Language
    {
        get { return language; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Язык не может быть пустым.");
            language = value;
        }
    }

    public Book(
        string title,
        string author,
        int yearOfPublication,
        string isbn = "Не указан",
        string publisher = "Не указан",
        int pages = 1,
        string language = "Русский")
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        Isbn = isbn;
        Publisher = publisher;
        Pages = pages;
        Language = language;
    }

    public override bool Equals(object obj)
    {
        if (obj is Book otherBook)
            return Isbn == otherBook.Isbn;
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Isbn);
    }

    public virtual void GetInfo()
    {
        Console.WriteLine(
            $"Название: {Title}; Автор: {Author}; Год: {YearOfPublication}; " +
            $"ISBN: {Isbn}; Издательство: {Publisher}; Страниц: {Pages}; Язык: {Language}");
    }

    public void GetInfo(bool detailed)
    {
        if (detailed)
            Console.WriteLine(GetFullDescription());
        else
            GetInfo();
    }

    public virtual void Read()
    {
        Console.WriteLine($"Чтение книги \"{Title}\"");
    }

    public virtual void Borrow()
    {
        Console.WriteLine($"Книга \"{Title}\" выдана на чтение");
    }

    public int GetAge()
    {
        return DateTime.Now.Year - YearOfPublication;
    }

    public bool IsModern()
    {
        return GetAge() <= 10;
    }

    public virtual string GetShortInfo()
    {
        return $"{Title} — {Author} ({YearOfPublication})";
    }

    public virtual string GetFullDescription()
    {
        return $"{GetShortInfo()}. Состояние: {Condition}. Инвентарный номер: {InventoryNumber}. Место: {ShelfLocation}. Добавлена: {DateAdded:dd.MM.yyyy}";
    }

    public double CalculateLateFee(int daysLate)
    {
        return daysLate * 10.0;
    }

    public void UpdateCondition(string newCondition)
    {
        Condition = newCondition;
        ConditionChanged?.Invoke($"Состояние книги \"{Title}\" изменено на: {newCondition}");
    }

    public override string ToString()
    {
        return GetShortInfo();
    }
}

public class Textbook : Book
{
    private string subject; private int gradeLevel; private bool hasExercises;
    public string MinistryApproval { get; set; } = "Отсутствует";
    public string Subject { get => subject; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); subject = value; } }
    public int GradeLevel { get => gradeLevel; set { if (value < 1 || value > 11) throw new ArgumentException(); gradeLevel = value; } }
    public bool HasExercises { get => hasExercises; set => hasExercises = value; }
    public Textbook(string title, string author, int yearOfPublication, string subject, int gradeLevel, bool hasExercises = true, string isbn = "Не указан", string publisher = "Не указан", int pages = 1, string language = "Русский") : base(title, author, yearOfPublication, isbn, publisher, pages, language) { Subject = subject; GradeLevel = gradeLevel; HasExercises = hasExercises; }
    public override void Read() => Console.WriteLine($"Чтение учебника \"{Title}\" по предмету {Subject}, {GradeLevel} класс");
    public override void GetInfo() => Console.WriteLine($"УЧЕБНИК: {Title}; Автор: {Author}; Год: {YearOfPublication}; Предмет: {Subject}; Класс: {GradeLevel}; Задания: {(HasExercises ? "есть" : "нет")}; ISBN: {Isbn}");
    public override string GetFullDescription() => $"{base.GetFullDescription()}. Предмет: {Subject}. Класс: {GradeLevel}. Гриф: {MinistryApproval}";
    public void SolveExercise(int number) => Console.WriteLine($"Решение упражнения №{number} из учебника \"{Title}\"");
    public void SolveExercise(int number, string difficulty) => Console.WriteLine($"Решение упражнения №{number} (сложность: {difficulty}) из учебника \"{Title}\"");
    public override string GetShortInfo() => $"{base.GetShortInfo()}, предмет: {Subject}, класс: {GradeLevel}";
}

public class ProgrammingTextbook : Textbook
{
    private string programmingLanguage; private string framework;
    public string ProgrammingLanguage { get => programmingLanguage; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); programmingLanguage = value; } }
    public string Framework { get => framework; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); framework = value; } }
    public ProgrammingTextbook(string title, string author, int yearOfPublication, string subject, int gradeLevel, string programmingLanguage, string framework, bool hasExercises = true, string isbn = "Не указан", string publisher = "Не указан", int pages = 1, string language = "Русский") : base(title, author, yearOfPublication, subject, gradeLevel, hasExercises, isbn, publisher, pages, language) { ProgrammingLanguage = programmingLanguage; Framework = framework; }
    public override void Read() => Console.WriteLine($"Чтение учебника по программированию \"{Title}\" ({ProgrammingLanguage}, {Framework})");
    public override void GetInfo() => Console.WriteLine($"УЧЕБНИК ПРОГРАММИРОВАНИЯ: {Title}; Автор: {Author}; Год: {YearOfPublication}; Предмет: {Subject}; Класс: {GradeLevel}; Язык: {ProgrammingLanguage}; Фреймворк: {Framework}");
    public void RunCodeExample() => Console.WriteLine($"Запуск примера кода на {ProgrammingLanguage} из книги \"{Title}\"");
}

public class Fiction : Book
{
    private string genre; private string mainCharacter; private int rating;
    public string LiteraryAward { get; set; } = "Нет";
    public string Genre { get => genre; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); genre = value; } }
    public string MainCharacter { get => mainCharacter; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); mainCharacter = value; } }
    public int Rating { get => rating; set { if (value < 0 || value > 5) throw new ArgumentException(); rating = value; } }
    public Fiction(string title, string author, int yearOfPublication, string genre, string mainCharacter, int rating = 5, string isbn = "Не указан", string publisher = "Не указан", int pages = 1, string language = "Русский") : base(title, author, yearOfPublication, isbn, publisher, pages, language) { Genre = genre; MainCharacter = mainCharacter; Rating = rating; }
    public override void Borrow() => Console.WriteLine($"Выдана художественная книга \"{Title}\" жанра {Genre}");
    public override void GetInfo() => Console.WriteLine($"ХУДОЖЕСТВЕННАЯ ЛИТЕРАТУРА: {Title}; Автор: {Author}; Год: {YearOfPublication}; Жанр: {Genre}; Главный герой: {MainCharacter}; Рейтинг: {Rating}/5");
    public override string GetFullDescription() => $"{base.GetFullDescription()}. Жанр: {Genre}. Награды: {LiteraryAward}";
    public void PrintQuote() => Console.WriteLine($"Цитата из книги \"{Title}\"...");
}

public class ScientificLiterature : Book
{
    private string fieldOfScience; private string doi; private int citationCount;
    public string PeerReviewStatus { get; set; } = "Не рецензировано";
    public string FieldOfScience { get => fieldOfScience; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); fieldOfScience = value; } }
    public string Doi { get => doi; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); doi = value; } }
    public int CitationCount { get => citationCount; set { if (value < 0) throw new ArgumentException(); citationCount = value; } }
    public ScientificLiterature(string title, string author, int yearOfPublication, string fieldOfScience, string doi = "Не указан", int citationCount = 0, string isbn = "Не указан", string publisher = "Не указан", int pages = 1, string language = "Русский") : base(title, author, yearOfPublication, isbn, publisher, pages, language) { FieldOfScience = fieldOfScience; Doi = doi; CitationCount = citationCount; }
    public override void GetInfo() => Console.WriteLine($"НАУЧНАЯ ЛИТЕРАТУРА: {Title}; Автор: {Author}; Год: {YearOfPublication}; Область науки: {FieldOfScience}; DOI: {Doi}; Цитирований: {CitationCount}");
    public override string GetFullDescription() => $"{base.GetFullDescription()}. Область: {FieldOfScience}. Статус рецензии: {PeerReviewStatus}";
    public void Cite() => Console.WriteLine($"Ссылка: {Author}. {Title}. {YearOfPublication}. DOI: {Doi}");
}

public interface IOnlineReadable { void OpenOnline(); }
public interface IAudioBook { void Listen(); }

public class InteractiveProgrammingTextbook : ProgrammingTextbook, IOnlineReadable, IAudioBook
{
    private string onlinePlatform;
    public string OnlinePlatform { get => onlinePlatform; set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException(); onlinePlatform = value; } }
    public InteractiveProgrammingTextbook(string title, string author, int yearOfPublication, string subject, int gradeLevel, string programmingLanguage, string framework, string onlinePlatform, bool hasExercises = true, string isbn = "Не указан", string publisher = "Не указан", int pages = 1, string language = "Русский") : base(title, author, yearOfPublication, subject, gradeLevel, programmingLanguage, framework, hasExercises, isbn, publisher, pages, language) { OnlinePlatform = onlinePlatform; }
    void IOnlineReadable.OpenOnline() => Console.WriteLine($"Открыт онлайн-доступ к \"{Title}\" на платформе {OnlinePlatform}");
    void IAudioBook.Listen() => Console.WriteLine($"Прослушивание аудиоверсии \"{Title}\"");
    public override void GetInfo() => Console.WriteLine($"ИНТЕРАКТИВНЫЙ УЧЕБНИК: {Title}; Автор: {Author}; Предмет: {Subject}; Язык: {ProgrammingLanguage}; Платформа: {OnlinePlatform}");
}

public interface IBookStorage
{
    void Add(Book book);
    bool Remove(Book book);
    bool Contains(Book book);
    IEnumerable<Book> GetAll();
    Book FindByTitle(string title);
}

public class InMemoryBookStorage : IBookStorage
{
    private readonly HashSet<Book> books = new HashSet<Book>();

    public void Add(Book book) => books.Add(book);
    public bool Remove(Book book) => books.Remove(book);
    public bool Contains(Book book) => books.Contains(book);
    public IEnumerable<Book> GetAll() => books.ToList();
    public Book FindByTitle(string title) => books.FirstOrDefault(b => b.Title.Equals(title, StringComparison.OrdinalIgnoreCase));
}

public interface ILibrary
{
    void AddBook(Book book);
    void AddBook(IEnumerable<Book> books);
    bool LendBook(Book book, Reader reader);
    void ReceiveBook(Book book, Reader reader);
    void ShowAvailableBooks();
}

public class Library : ILibrary
{
    private readonly IBookStorage storage;
    private readonly Dictionary<Book, Reader> issuedBooks = new Dictionary<Book, Reader>();

    public delegate void LibraryNotification(string message);
    public event LibraryNotification BookIssued;
    public event LibraryNotification BookReturned;

    public Library(IBookStorage storage)
    {
        this.storage = storage;
    }

    public void AddBook(Book book)
    {
        storage.Add(book);
        Console.WriteLine($"Книга {book.Title} добавлена в библиотеку.");
    }

    public void AddBook(IEnumerable<Book> newBooks)
    {
        foreach (var book in newBooks) AddBook(book);
    }

    public bool LendBook(Book book, Reader reader)
    {
        if (storage.Contains(book))
        {
            storage.Remove(book);
            issuedBooks[book] = reader;
            BookIssued?.Invoke($"Книга \"{book.Title}\" выдана читателю {reader.Name}.");
            return true;
        }
        Console.WriteLine($"Книга {book.Title} недоступна для выдачи.");
        return false;
    }

    public void ReceiveBook(Book book, Reader reader)
    {
        if (issuedBooks.ContainsKey(book) && issuedBooks[book] == reader)
        {
            issuedBooks.Remove(book);
            storage.Add(book);
            BookReturned?.Invoke($"Книга \"{book.Title}\" возвращена читателем {reader.Name}.");
        }
        else
        {
            Console.WriteLine($"Ошибка при возврате книги {book.Title}.");
        }
    }

    public void ShowAvailableBooks()
    {
        Console.WriteLine("Доступные книги в библиотеке:");
        foreach (var book in storage.GetAll()) book.GetInfo();
    }
}

public class Reader
{
    public string Name { get; }
    private List<Book> borrowedBooks = new List<Book>();
    public IReadOnlyList<Book> BorrowedBooks => borrowedBooks;

    public Reader(string name)
    {
        if (string.IsNullOrWhiteSpace(name))
            throw new ArgumentException("Имя читателя не может быть пустым.");
        Name = name;
    }

    public void TakeBook(Book book, ILibrary library)
    {
        if (library.LendBook(book, this)) borrowedBooks.Add(book);
    }

    public void ReadBook(Book book)
    {
        if (borrowedBooks.Contains(book)) book.Read();
        else Console.WriteLine($"{Name} не взял книгу {book.Title} для чтения.");
    }

    public void ReturnBook(Book book, ILibrary library)
    {
        if (borrowedBooks.Contains(book))
        {
            borrowedBooks.Remove(book);
            library.ReceiveBook(book, this);
        }
        else Console.WriteLine($"{Name} не может вернуть книгу {book.Title}, так как она не была взята.");
    }
}

public class Catalog<T> where T : Book
{
    private List<T> items = new List<T>();
    public void AddItem(T item) => items.Add(item);
    public T FindByTitle(string title) => items.FirstOrDefault(i => i.Title.Equals(title, StringComparison.OrdinalIgnoreCase));
    public void ShowAll()
    {
        Console.WriteLine("Содержимое каталога:");
        foreach (var item in items) Console.WriteLine(item.GetShortInfo());
    }
}

public static class Program
{
    static void OnBookConditionChanged(string message)
    {
        Console.WriteLine($"[Событие книги] {message}");
    }

    public static void Main()
    {
        IBookStorage storage = new InMemoryBookStorage();
        ILibrary library = new Library(storage);

        Book wap = new Book("Война и мир", "Л.Н. Толстой", 1869, "978-5-17-090000-0", "Русский вестник", 1300, "Русский");
        wap.InventoryNumber = 101; wap.ShelfLocation = "A-1";

        Textbook alg = new Textbook("Алгебра 9 класс", "Макарычев Ю.Н.", 2020, "Математика", 9, true, "978-5-09-000000-1", "Просвещение", 250);
        alg.InventoryNumber = 102; alg.ShelfLocation = "B-2"; alg.MinistryApproval = "Рекомендовано Минобрнауки";

        Fiction mas = new Fiction("Мастер и Маргарита", "М.А. Булгаков", 1967, "Роман", "Воланд", 5, "978-5-17-000000-2", "Художественная литература", 480);
        mas.InventoryNumber = 103; mas.LiteraryAward = "Букер";

        ScientificLiterature lst = new ScientificLiterature("Краткая история времени", "С. Хокинг", 1988, "Физика", "10.1000/abc", 5000, "978-5-03-000000-3", "Мир", 256);
        lst.InventoryNumber = 104; lst.PeerReviewStatus = "Прошло рецензирование";

        library.AddBook(wap);
        library.AddBook(alg);
        library.AddBook(mas);
        library.AddBook(lst);

        Console.WriteLine("\n=== Проверка уникальности HashSet ===");
        Book duplicateWap = new Book("Война и мир (копия)", "Л.Н. Толстой", 1869, "978-5-17-090000-0", "Русский вестник", 1300, "Русский");
        library.AddBook(duplicateWap);
        Console.WriteLine($"Количество уникальных книг в фонде: {storage.GetAll().Count()}");

        Console.WriteLine("\n=== События ===");
        
        wap.ConditionChanged += OnBookConditionChanged;

        ((Library)library).BookIssued += msg => Console.WriteLine($"[Уведомление] {msg}");
        ((Library)library).BookReturned += msg => Console.WriteLine($"[Уведомление] {msg}");

        wap.UpdateCondition("Требует ремонта");

        Reader ivan = new Reader("Иван");
        ivan.TakeBook(wap, library);
        ivan.ReturnBook(wap, library);

        Console.WriteLine("\n=== Перегрузка в учебниках ===");
        alg.SolveExercise(15);
        alg.SolveExercise(15, "Повышенная");
    }
}

Program.Main();
```

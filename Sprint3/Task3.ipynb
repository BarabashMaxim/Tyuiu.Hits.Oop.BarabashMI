```
using System;
using System.Collections.Generic;
using System.Linq;

public class Book
{
    private string title;
    private string author;
    private int yearOfPublication;
    private string isbn;
    private string publisher;
    private int pages;
    private string language;
    
    public string Condition { get; set; } = "Хорошее";
    public int InventoryNumber { get; set; }
    public string ShelfLocation { get; set; } = "Не указано";
    public DateTime DateAdded { get; set; } = DateTime.Now;

    public string Title
    {
        get { return title; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название книги не может быть пустым.");
            title = value;
        }
    }

    public string Author
    {
        get { return author; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Автор книги не может быть пустым.");
            author = value;
        }
    }

    public int YearOfPublication
    {
        get { return yearOfPublication; }
        set
        {
            if (value <= 0 || value > DateTime.Now.Year)
                throw new ArgumentException("Некорректный год публикации.");
            yearOfPublication = value;
        }
    }

    public string Isbn
    {
        get { return isbn; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("ISBN не может быть пустым.");
            isbn = value;
        }
    }

    public string Publisher
    {
        get { return publisher; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Издательство не может быть пустым.");
            publisher = value;
        }
    }

    public int Pages
    {
        get { return pages; }
        set
        {
            if (value <= 0)
                throw new ArgumentException("Количество страниц должно быть больше 0.");
            pages = value;
        }
    }

    public string Language
    {
        get { return language; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Язык не может быть пустым.");
            language = value;
        }
    }

    public Book(
        string title,
        string author,
        int yearOfPublication,
        string isbn = "Не указан",
        string publisher = "Не указан",
        int pages = 1,
        string language = "Русский")
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        Isbn = isbn;
        Publisher = publisher;
        Pages = pages;
        Language = language;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine(
            $"Название: {Title}; Автор: {Author}; Год: {YearOfPublication}; " +
            $"ISBN: {Isbn}; Издательство: {Publisher}; Страниц: {Pages}; Язык: {Language}");
    }

    public void GetInfo(bool detailed)
    {
        if (detailed)
        {
            Console.WriteLine(GetFullDescription());
        }
        else
        {
            GetInfo();
        }
    }

    public virtual void Read()
    {
        Console.WriteLine($"Чтение книги \"{Title}\"");
    }

    public virtual void Borrow()
    {
        Console.WriteLine($"Книга \"{Title}\" выдана на чтение");
    }

    public int GetAge()
    {
        return DateTime.Now.Year - YearOfPublication;
    }

    public bool IsModern()
    {
        return GetAge() <= 10;
    }

    public virtual string GetShortInfo()
    {
        return $"{Title} — {Author} ({YearOfPublication})";
    }

    public virtual string GetFullDescription()
    {
        return $"{GetShortInfo()}. Состояние: {Condition}. Инвентарный номер: {InventoryNumber}. Место: {ShelfLocation}. Добавлена: {DateAdded:dd.MM.yyyy}";
    }

    public double CalculateLateFee(int daysLate)
    {
        return daysLate * 10.0;
    }

    public void UpdateCondition(string newCondition)
    {
        Condition = newCondition;
    }

    public override string ToString()
    {
        return GetShortInfo();
    }
}

public class Textbook : Book
{
    private string subject;
    private int gradeLevel;
    private bool hasExercises;
    
    public string MinistryApproval { get; set; } = "Отсутствует";

    public string Subject
    {
        get { return subject; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Предмет учебника не может быть пустым.");
            subject = value;
        }
    }

    public int GradeLevel
    {
        get { return gradeLevel; }
        set
        {
            if (value < 1 || value > 11)
                throw new ArgumentException("Класс должен быть от 1 до 11.");
            gradeLevel = value;
        }
    }

    public bool HasExercises
    {
        get { return hasExercises; }
        set { hasExercises = value; }
    }

    public Textbook(
        string title,
        string author,
        int yearOfPublication,
        string subject,
        int gradeLevel,
        bool hasExercises = true,
        string isbn = "Не указан",
        string publisher = "Не указан",
        int pages = 1,
        string language = "Русский")
        : base(title, author, yearOfPublication, isbn, publisher, pages, language)
    {
        Subject = subject;
        GradeLevel = gradeLevel;
        HasExercises = hasExercises;
    }

    public override void Read()
    {
        Console.WriteLine($"Чтение учебника \"{Title}\" по предмету {Subject}, {GradeLevel} класс");
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"УЧЕБНИК: {Title}; Автор: {Author}; Год: {YearOfPublication}; " +
            $"Предмет: {Subject}; Класс: {GradeLevel}; " +
            $"Задания: {(HasExercises ? "есть" : "нет")}; ISBN: {Isbn}");
    }

    public override string GetFullDescription()
    {
        return $"{base.GetFullDescription()}. Предмет: {Subject}. Класс: {GradeLevel}. Гриф: {MinistryApproval}";
    }

    public void SolveExercise(int number)
    {
        Console.WriteLine($"Решение упражнения №{number} из учебника \"{Title}\"");
    }

    public void SolveExercise(int number, string difficulty)
    {
        Console.WriteLine($"Решение упражнения №{number} (сложность: {difficulty}) из учебника \"{Title}\"");
    }

    public override string GetShortInfo()
    {
        return $"{base.GetShortInfo()}, предмет: {Subject}, класс: {GradeLevel}";
    }
}

public class ProgrammingTextbook : Textbook
{
    private string programmingLanguage;
    private string framework;

    public string ProgrammingLanguage
    {
        get { return programmingLanguage; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Язык программирования не может быть пустым.");
            programmingLanguage = value;
        }
    }

    public string Framework
    {
        get { return framework; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Фреймворк не может быть пустым.");
            framework = value;
        }
    }

    public ProgrammingTextbook(
        string title,
        string author,
        int yearOfPublication,
        string subject,
        int gradeLevel,
        string programmingLanguage,
        string framework,
        bool hasExercises = true,
        string isbn = "Не указан",
        string publisher = "Не указан",
        int pages = 1,
        string language = "Русский")
        : base(title, author, yearOfPublication, subject, gradeLevel, hasExercises, isbn, publisher, pages, language)
    {
        ProgrammingLanguage = programmingLanguage;
        Framework = framework;
    }

    public override void Read()
    {
        Console.WriteLine(
            $"Чтение учебника по программированию \"{Title}\" " +
            $"({ProgrammingLanguage}, {Framework})");
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"УЧЕБНИК ПРОГРАММИРОВАНИЯ: {Title}; Автор: {Author}; Год: {YearOfPublication}; " +
            $"Предмет: {Subject}; Класс: {GradeLevel}; Язык: {ProgrammingLanguage}; " +
            $"Фреймворк: {Framework}");
    }

    public void RunCodeExample()
    {
        Console.WriteLine($"Запуск примера кода на {ProgrammingLanguage} из книги \"{Title}\"");
    }
}

public class Fiction : Book
{
    private string genre;
    private string mainCharacter;
    private int rating;
    
    public string LiteraryAward { get; set; } = "Нет";

    public string Genre
    {
        get { return genre; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Жанр не может быть пустым.");
            genre = value;
        }
    }

    public string MainCharacter
    {
        get { return mainCharacter; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Главный герой не может быть пустым.");
            mainCharacter = value;
        }
    }

    public int Rating
    {
        get { return rating; }
        set
        {
            if (value < 0 || value > 5)
                throw new ArgumentException("Рейтинг должен быть от 0 до 5.");
            rating = value;
        }
    }

    public Fiction(
        string title,
        string author,
        int yearOfPublication,
        string genre,
        string mainCharacter,
        int rating = 5,
        string isbn = "Не указан",
        string publisher = "Не указан",
        int pages = 1,
        string language = "Русский")
        : base(title, author, yearOfPublication, isbn, publisher, pages, language)
    {
        Genre = genre;
        MainCharacter = mainCharacter;
        Rating = rating;
    }

    public override void Borrow()
    {
        Console.WriteLine($"Выдана художественная книга \"{Title}\" жанра {Genre}");
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"ХУДОЖЕСТВЕННАЯ ЛИТЕРАТУРА: {Title}; Автор: {Author}; Год: {YearOfPublication}; " +
            $"Жанр: {Genre}; Главный герой: {MainCharacter}; Рейтинг: {Rating}/5");
    }

    public override string GetFullDescription()
    {
        return $"{base.GetFullDescription()}. Жанр: {Genre}. Награды: {LiteraryAward}";
    }

    public void PrintQuote()
    {
        Console.WriteLine($"Цитата из книги \"{Title}\"...");
    }
}

public class ScientificLiterature : Book
{
    private string fieldOfScience;
    private string doi;
    private int citationCount;
    
    public string PeerReviewStatus { get; set; } = "Не рецензировано";

    public string FieldOfScience
    {
        get { return fieldOfScience; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Область науки не может быть пустой.");
            fieldOfScience = value;
        }
    }

    public string Doi
    {
        get { return doi; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("DOI не может быть пустым.");
            doi = value;
        }
    }

    public int CitationCount
    {
        get { return citationCount; }
        set
        {
            if (value < 0)
                throw new ArgumentException("Количество цитирований не может быть отрицательным.");
            citationCount = value;
        }
    }

    public ScientificLiterature(
        string title,
        string author,
        int yearOfPublication,
        string fieldOfScience,
        string doi = "Не указан",
        int citationCount = 0,
        string isbn = "Не указан",
        string publisher = "Не указан",
        int pages = 1,
        string language = "Русский")
        : base(title, author, yearOfPublication, isbn, publisher, pages, language)
    {
        FieldOfScience = fieldOfScience;
        Doi = doi;
        CitationCount = citationCount;
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"НАУЧНАЯ ЛИТЕРАТУРА: {Title}; Автор: {Author}; Год: {YearOfPublication}; " +
            $"Область науки: {FieldOfScience}; DOI: {Doi}; Цитирований: {CitationCount}");
    }

    public override string GetFullDescription()
    {
        return $"{base.GetFullDescription()}. Область: {FieldOfScience}. Статус рецензии: {PeerReviewStatus}";
    }

    public void Cite()
    {
        Console.WriteLine($"Ссылка: {Author}. {Title}. {YearOfPublication}. DOI: {Doi}");
    }
}

public interface IOnlineReadable
{
    void OpenOnline();
}

public interface IAudioBook
{
    void Listen();
}

public class InteractiveProgrammingTextbook : ProgrammingTextbook, IOnlineReadable, IAudioBook
{
    private string onlinePlatform;

    public string OnlinePlatform
    {
        get { return onlinePlatform; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Онлайн-платформа не может быть пустой.");
            onlinePlatform = value;
        }
    }

    public InteractiveProgrammingTextbook(
        string title,
        string author,
        int yearOfPublication,
        string subject,
        int gradeLevel,
        string programmingLanguage,
        string framework,
        string onlinePlatform,
        bool hasExercises = true,
        string isbn = "Не указан",
        string publisher = "Не указан",
        int pages = 1,
        string language = "Русский")
        : base(title, author, yearOfPublication, subject, gradeLevel,
               programmingLanguage, framework, hasExercises, isbn, publisher, pages, language)
    {
        OnlinePlatform = onlinePlatform;
    }

    public void OpenOnline()
    {
        Console.WriteLine($"Открыт онлайн-доступ к \"{Title}\" на платформе {OnlinePlatform}");
    }

    public void Listen()
    {
        Console.WriteLine($"Прослушивание аудиоверсии \"{Title}\"");
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"ИНТЕРАКТИВНЫЙ УЧЕБНИК: {Title}; Автор: {Author}; Предмет: {Subject}; " +
            $"Язык: {ProgrammingLanguage}; Платформа: {OnlinePlatform}");
    }
}

public class Reader
{
    public string Name { get; }

    private List<Book> borrowedBooks = new List<Book>();
    public IReadOnlyList<Book> BorrowedBooks => borrowedBooks;

    public Reader(string name)
    {
        if (string.IsNullOrWhiteSpace(name))
            throw new ArgumentException("Имя читателя не может быть пустым.");
        Name = name;
    }

    public void TakeBook(Book book, Library library)
    {
        if (library.LendBook(book, this))
        {
            borrowedBooks.Add(book);
            Console.WriteLine($"{Name} взял книгу {book.Title}");
        }
        else
        {
            Console.WriteLine($"{Name} не смог взять книгу {book.Title}");
        }
    }

    public void ReadBook(Book book)
    {
        if (borrowedBooks.Contains(book))
            book.Read();
        else
            Console.WriteLine($"{Name} не взял книгу {book.Title} для чтения.");
    }

    public void ReturnBook(Book book, Library library)
    {
        if (borrowedBooks.Contains(book))
        {
            borrowedBooks.Remove(book);
            library.ReceiveBook(book, this);
            Console.WriteLine($"{Name} вернул книгу {book.Title}");
        }
        else
        {
            Console.WriteLine($"{Name} не может вернуть книгу {book.Title}, так как она не была взята.");
        }
    }
}

public class Library
{
    private List<Book> books = new List<Book>();
    private Dictionary<Book, Reader> issuedBooks = new Dictionary<Book, Reader>();

    public void AddBook(Book book)
    {
        books.Add(book);
        Console.WriteLine($"Книга {book.Title} добавлена в библиотеку.");
    }

    public void AddBook(IEnumerable<Book> newBooks)
    {
        foreach (var book in newBooks)
        {
            AddBook(book);
        }
    }

    public bool LendBook(Book book, Reader reader)
    {
        if (books.Contains(book))
        {
            books.Remove(book);
            issuedBooks[book] = reader;
            Console.WriteLine($"Книга {book.Title} выдана читателю {reader.Name}.");
            return true;
        }

        Console.WriteLine($"Книга {book.Title} недоступна для выдачи.");
        return false;
    }

    public void ReceiveBook(Book book, Reader reader)
    {
        if (issuedBooks.ContainsKey(book) && issuedBooks[book] == reader)
        {
            issuedBooks.Remove(book);
            books.Add(book);
            Console.WriteLine($"Книга {book.Title} возвращена в библиотеку.");
        }
        else
        {
            Console.WriteLine($"Ошибка при возврате книги {book.Title}.");
        }
    }

    public void ShowAvailableBooks()
    {
        Console.WriteLine("Доступные книги в библиотеке:");
        foreach (var book in books)
        {
            book.GetInfo();
        }
    }
}

public class Catalog<T> where T : Book
{
    private List<T> items = new List<T>();

    public void AddItem(T item)
    {
        items.Add(item);
    }

    public T FindByTitle(string title)
    {
        return items.FirstOrDefault(i => i.Title.Equals(title, StringComparison.OrdinalIgnoreCase));
    }

    public void ShowAll()
    {
        Console.WriteLine("Содержимое каталога:");
        foreach (var item in items)
        {
            Console.WriteLine(item.GetShortInfo());
        }
    }
}

public static class Program
{
    public static void Main()
    {
        Library library = new Library();

        Book wap = new Book(
            "Война и мир", "Л.Н. Толстой", 1869,
            "978-5-17-090000-0", "Русский вестник", 1300, "Русский");
        wap.InventoryNumber = 101;
        wap.ShelfLocation = "A-1";

        Textbook alg = new Textbook(
            "Алгебра 9 класс", "Макарычев Ю.Н.", 2020,
            "Математика", 9, true,
            "978-5-09-000000-1", "Просвещение", 250);
        alg.InventoryNumber = 102;
        alg.ShelfLocation = "B-2";
        alg.MinistryApproval = "Рекомендовано Минобрнауки";

        Fiction mas = new Fiction(
            "Мастер и Маргарита", "М.А. Булгаков", 1967,
            "Роман", "Воланд", 5,
            "978-5-17-000000-2", "Художественная литература", 480);
        mas.InventoryNumber = 103;
        mas.LiteraryAward = "Букер";

        ScientificLiterature lst = new ScientificLiterature(
            "Краткая история времени", "С. Хокинг", 1988,
            "Физика", "10.1000/abc", 5000,
            "978-5-03-000000-3", "Мир", 256);
        lst.InventoryNumber = 104;
        lst.PeerReviewStatus = "Прошло рецензирование";

        ProgrammingTextbook csharp = new ProgrammingTextbook(
            "C# для начинающих", "Иванов И.И.", 2024,
            "Информатика", 10, "C#", ".NET",
            true, "978-5-00-000000-4", "Наука", 300);
        csharp.InventoryNumber = 105;

        InteractiveProgrammingTextbook interactive = new InteractiveProgrammingTextbook(
            "C# и .NET", "Петров П.П.", 2025,
            "Информатика", 11, "C#", ".NET", "Stepik",
            true, "978-5-00-000000-5", "Наука", 350);
        interactive.InventoryNumber = 106;

        library.AddBook(wap);
        library.AddBook(alg);
        library.AddBook(mas);
        library.AddBook(lst);
        library.AddBook(csharp);
        library.AddBook(interactive);

        Console.WriteLine("=== Доступные книги ===");
        library.ShowAvailableBooks();

        Book[] books = { wap, alg, mas, lst, csharp, interactive };

        Console.WriteLine("\n=== Демонстрация перегрузки и перекрытия ===");
        foreach (Book b in books)
        {
            b.GetInfo();
            b.GetInfo(true);
            b.Read();
            b.Borrow();
            Console.WriteLine($"Возраст: {b.GetAge()}, современная: {b.IsModern()}");
            Console.WriteLine($"Штраф за 5 дней просрочки: {b.CalculateLateFee(5)} руб.");
            Console.WriteLine();
        }

        Console.WriteLine("=== Перегрузка в учебниках ===");
        alg.SolveExercise(15);
        alg.SolveExercise(15, "Повышенная");

        Console.WriteLine("\n=== Читатели ===");
        Reader ivan = new Reader("Иван");
        Reader anna = new Reader("Анна");

        ivan.TakeBook(wap, library);
        ivan.ReadBook(wap);
        anna.TakeBook(wap, library);
        ivan.ReturnBook(wap, library);
        anna.TakeBook(wap, library);
        anna.ReadBook(wap);

        Console.WriteLine("\n");
        Catalog<Book> generalCatalog = new Catalog<Book>();
        generalCatalog.AddItem(wap);
        generalCatalog.AddItem(alg);
        generalCatalog.AddItem(mas);
        generalCatalog.AddItem(lst);
        generalCatalog.AddItem(csharp);
        generalCatalog.AddItem(interactive);
        generalCatalog.ShowAll();

        Book found = generalCatalog.FindByTitle("Война и мир");
        if (found != null)
        {
            Console.WriteLine($"Найдена книга: {found.GetShortInfo()}");
        }
        Console.WriteLine("\n");
        interactive.OpenOnline();
        interactive.Listen();
        interactive.GetInfo();
        
        library.AddBook(new Book[] 
        { 
            new Book("Тестовая книга 1", "Автор 1", 2000), 
            new Book("Тестовая книга 2", "Автор 2", 2001) 
        });
    }
}

Program.Main();
```
